# Midpoint rule for the sine integral

The integral of sin(x) from 0 to pi is 2, so the midpoint rule gives a check.
This tiny problem exists only to test the tooling.

```python
from math import pi, sin


def midpoint(a, b, n):
    """Midpoint-rule estimate of the integral of sin(x) on [a, b]."""
    h = (b - a) / n  # the width of one subinterval
    integral_sum = 0.0
    for i in range(n):
        integral_sum += sin(a + (i + 0.5) * h)
    return integral_sum * h
```

```python
import math

upper = pi  # the end of the interval
approx = midpoint(0.0, upper, 1000)
print('approx =', approx, 'error =', abs(approx - 2.0), math.pi)
```
