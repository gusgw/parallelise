# Midpoint rule for the sine integral

The integral of sin(x) from 0 to pi is 2, so the midpoint rule gives a check. This small
problem exists only to test the tool.

```python
import numpy as np


def midpoint(a, b, n):
    """Midpoint-rule estimate of the integral of sin(x) on [a, b] with n subintervals."""
    h = (b - a) / n
    return np.sum(np.sin(a + (np.arange(n) + 0.5) * h)) * h


def make_inputs(size, rng):
    return {"a": rng.uniform(-1.0, 0.0), "b": rng.uniform(2.0, 3.0), "n": size}


ROUTINE = {
    "function": midpoint,
    "make_inputs": make_inputs,
    "arguments": {"a": "float64", "b": "float64", "n": "int64"},
    "results": {"value": "float64"},
    "test_sizes": [1, 1_000, 100_000],
    "benchmark_size": 10_000_000,
    "tolerance": 1e-8,
}
```
