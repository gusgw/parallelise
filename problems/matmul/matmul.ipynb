# Dense matrix multiply (research prototype)

Compute `C = A @ B` for two `n x n` matrices. `A` and `B` use a fixed deterministic
integer fill so the result is reproducible; we report the checksum (sum of all entries
of `C`) as the oracle. The C port must implement `C = A*B` and match this checksum.

```python
N = 256

def A_of(i, j):
    return ((i * 3 + j * 7) % 11) + 1

def B_of(i, j):
    return ((i * 5 + j * 2) % 13) + 1

A = [[A_of(i, j) for j in range(N)] for i in range(N)]
B = [[B_of(i, j) for j in range(N)] for i in range(N)]
```

```python
# naive triple-loop matrix multiply
def matmul(A, B, n):
    C = [[0.0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            s = 0.0
            for k in range(n):
                s += A[i][k] * B[k][j]
            C[i][j] = s
    return C

C = matmul(A, B, N)
checksum = sum(C[i][j] for i in range(N) for j in range(N))
print('checksum =', checksum)   # -> 704626403
```

```python
# The oracle: this notebook's own output. A port of the algorithm (in any
# language) must reproduce this checksum within tol. The individual entries of C
# are handed over as the terms of the checksum, so the extraction tool can measure
# how much the sum legitimately moves under reordering (here: not at all -- the
# entries are exact integers in double precision).
ORACLE = {'kernel': 'matmul',
          'params': {'N': N},
          'expected': {'CHECKSUM': float(checksum)},
          'terms': {'CHECKSUM': [C[i][j] for i in range(N) for j in range(N)]},
          'tol': 0.5}
```
