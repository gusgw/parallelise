# Product of two square matrices

The product `C = A B` of two `n` by `n` matrices has the entry `C[i, j]` equal to the sum, over
`k`, of `A[i, k]` times `B[k, j]`. The work grows as the cube of `n` while the data grow only as
its square, so every number is used many times.

```python
import numpy as np


def matmul(A, B):
    """The matrix product C = A B of two square matrices."""
    return A @ B
```

```python
print(matmul(np.array([[1.0, 2.0], [3.0, 4.0]]), np.array([[0.0, 1.0], [1.0, 0.0]])))
```

The declaration below tells the harness how to call the function and how to check a routine
against it. `make_inputs` fills both matrices with random numbers between -1 and 1, and sets a
few entries to exactly -1, 0 or 1. The tolerance is 3e-14 of the largest entry of the product,
which is ten times the difference of about 3e-15 between the product that NumPy computes and the
sum of products that a plain C loop computes.

```python
def make_inputs(size, rng):
    matrices = {}
    for name in ("A", "B"):
        matrix = rng.uniform(-1.0, 1.0, (size, size))
        special = rng.random((size, size)) < 0.05
        matrix[special] = rng.choice([-1.0, 0.0, 1.0], special.sum())
        matrices[name] = matrix
    return matrices


ROUTINE = {
    "function": matmul,
    "make_inputs": make_inputs,
    "arguments": {"A": "float64[n, n]", "B": "float64[n, n]"},
    "results": {"C": "float64[n, n]"},
    "test_sizes": [1, 17, 100, 300],
    "benchmark_size": 1024,
    "tolerance": 3e-14,
}
```
