# Estimating pi by numerical integration

Since the integral of 4/(1+x²) from 0 to 1 is pi, the midpoint rule over [0, 1] gives an
estimate of pi. The function below applies the rule on any interval.

```python
import numpy as np


def integrate(a, b, n):
    """Midpoint-rule estimate of the integral of 4/(1+x**2) on [a, b] with n subintervals."""
    h = (b - a) / n
    x = a + (np.arange(n) + 0.5) * h
    return np.sum(4.0 / (1.0 + x * x)) * h
```

```python
print(integrate(0.0, 1.0, 1_000_000), "should be close to", np.pi)
```

The declaration below tells the harness how to call the function and how to check a routine
against it. `make_inputs` draws the ends of the interval at random, so that every call
integrates over a different range. The tolerance, 1e-11 of the result, is about twenty times
the largest difference that honest orders of addition make at the benchmark size: 5.3e-13,
measured over twenty random intervals for the plain loop, NumPy's pairwise sum, contiguous
blocks for up to 64 workers and the strided pattern of a graphics card.

```python
def make_inputs(size, rng):
    return {"a": rng.uniform(-1.0, 0.0), "b": rng.uniform(1.0, 2.0), "n": size}


ROUTINE = {
    "function": integrate,
    "make_inputs": make_inputs,
    "arguments": {"a": "float64", "b": "float64", "n": "int64"},
    "results": {"area": "float64"},
    "test_sizes": [1, 1_000, 100_000],
    "benchmark_size": 50_000_000,
    "tolerance": 1e-11,
}
```
