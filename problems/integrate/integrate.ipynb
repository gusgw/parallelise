# Estimating π by numerical integration

Since ∫₀¹ 4/(1+x²) dx = π, the midpoint rule over [0,1] gives an
estimate of π. Research-quality Python below; the task is to turn it into
modular, unit-tested, parallel C with performance instrumentation.

```python
def integrate(a, b, n):
    """Midpoint-rule estimate of the integral of 4/(1+x**2) on [a, b]."""
    h = (b - a) / n
    s = 0.0
    for i in range(n):
        x = a + (i + 0.5) * h
        s += 4.0 / (1.0 + x * x)
    return s * h
```

```python
import math
approx = integrate(0.0, 1.0, 1_000_000)
print('approx =', approx, '  error =', abs(approx - math.pi))
```
