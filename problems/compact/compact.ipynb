# Selecting the elements above a threshold

Given an array of numbers and a threshold, this finds the positions of the numbers that are
greater than the threshold, in increasing order. The number of positions found depends on the
data, so the result is an array whose length is not known in advance. A serial program writes
each position it finds into the next free place of the output. Several workers that share the
task cannot do this, because each worker must first learn how many positions the workers before
it will find.

```python
import numpy as np


def compact(x, threshold):
    """The indices, in increasing order, of the elements of `x` that exceed `threshold`."""
    return np.flatnonzero(x > threshold)
```

```python
print(compact(np.array([0.2, 0.9, 0.5, 0.7, 0.1, 0.8]), 0.6))
```

The declaration below tells the harness how to call the function and how to check a routine
against it. The result is written `int64[<=n]` because it has at most `n` elements. The C
routine receives an array with room for `n` of them, and it also receives a pointer to a number
in which it must write how many it has filled in. The indices are whole numbers, so a routine
must reproduce them exactly. `make_inputs` draws random numbers between 0 and 1 and a random
threshold, so that any share of the elements may be selected. Now and then the threshold is such
that nothing is selected or everything is, and a few elements are set equal to the threshold,
which must not be selected.

```python
def make_inputs(size, rng):
    x = rng.random(size)
    threshold = float(rng.choice([-1.0, 1.0, rng.random()], p=[0.05, 0.05, 0.9]))
    x[rng.random(size) < 0.05] = threshold
    return {"x": x, "threshold": threshold}


ROUTINE = {
    "function": compact,
    "make_inputs": make_inputs,
    "arguments": {"x": "float64[n]", "threshold": "float64"},
    "results": {"indices": "int64[<=n]"},
    "test_sizes": [1, 1_000, 1_000_000],
    "benchmark_size": 60_000_000,
    "tolerance": 0,
}
```
